```python
import pandas as pd
import os
from sklearn.model_selection import train_test_split

# Load the dataset
df = pd.read_csv('mbti_1.csv')
print(f"Total samples: {len(df)}")
df['IE'] = df['type'].str[0]  # Introversion / Extroversion
df['NS'] = df['type'].str[1]  # Intuition / Sensing
df['TF'] = df['type'].str[2]  # Thinking / Feeling
df['JP'] = df['type'].str[3]  # Judging / Perceiving

print("New columns added:")
print(df[['type', 'IE', 'NS', 'TF', 'JP']].head())
df.head()
```

```python
# First split: 60% for training and 40% for the temporary set (val + test)
train_df, temp_df = train_test_split(
    df, 
    test_size=0.40, 
    random_state=42, 
    stratify=df['type']
)

# Second split: Divide the 40% temporary set equally into validation and test (20% each of the total)
val_df, test_df = train_test_split(
    temp_df, 
    test_size=0.50, 
    random_state=42, 
    stratify=temp_df['type']
)

print(f"Train size: {len(train_df)} (60%)")
print(f"Val size:   {len(val_df)} (20%)")
print(f"Test size:  {len(test_df)} (20%)")
```

```python
# Define output directory
output_dir = '../../split/mbti'
os.makedirs(output_dir, exist_ok=True)

# Save to CSV
train_df.to_csv(os.path.join(output_dir, 'train.csv'), index=False)
val_df.to_csv(os.path.join(output_dir, 'val.csv'), index=False)
test_df.to_csv(os.path.join(output_dir, 'test.csv'), index=False)

print(f"Files saved successfully in {output_dir}")
```
