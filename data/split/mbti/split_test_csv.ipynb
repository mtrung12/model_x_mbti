# Split `test.csv` into 10 parts

This notebook reads `test.csv` from the current folder and writes `test1.csv` through `test10.csv` alongside it.

If the row count is not divisible by 10, it distributes the extra rows across the first few files.

```python
from pathlib import Path

import numpy as np
import pandas as pd

data_dir = Path.cwd()
input_path = data_dir / "test.csv"

df = pd.read_csv(input_path)
n_parts = 10

parts = np.array_split(df, n_parts)

for idx, part_df in enumerate(parts, start=1):
    output_path = data_dir / f"test{idx}.csv"
    part_df.to_csv(output_path, index=False)
    print(f"Saved {output_path.name} with {len(part_df)} rows")

print("Done.")
```
